```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("seaborn_student_performance_dataset.csv")
```

```python
sns.barplot(data=df, x="Branch", y="Python") 
plt.title("Average Python Marks by Branch") 
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4.2))    #Creates a figure and one graph (axis).
sns.histplot(df["Python"], kde=True, color="#2E86C1", ax=ax)    #Creates a histogram of Python marks.
ax.set_title("Distribution of Python Marks with KDE Curve") #Sets the title of the graph.
plt.show()  #Displays the graph on the screen.
```

```python
sns.boxplot(data=df, x="Branch", y="Python", color="lightpink")
sns.violinplot(data=df, x="Branch", y="Python", color="skyblue")
plt.title("Distribution of Python Marks by Branch")
plt.show()  # Displays the graph on the screen.
```

```python
fig, ax = plt.subplots(figsize=(7, 4.2))
colors = ["skyblue", "lightpink"]
sns.scatterplot(
    data=df,
    x="StudyHours",
    y="Python",
    hue="Gender",
    style="Branch",
    palette=colors,
    s=100,
    ax=ax
)
ax.set_title("Study Hours vs Python Marks")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Python Marks")
plt.show()
```
