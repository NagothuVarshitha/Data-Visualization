```python
from google import genai
import getpass
```

```python
from dotenv import load_dotenv
import os
load_dotenv()  # Load environment variables from .env file
api_key = os.getenv("api_key")  # Retrieve the API key from environment variables
print(api_key[:5])  # Print the first 5 characters of the API key for verification
 
```

```python
from google import genai
client = genai.Client(
    api_key=api_key
)
```

```python
for model in client.models.list():
    print(model.name)
```

```python
response = client.models.generate_content(
    model="models/gemini-3.5-flash-lite",
    contents="Explain Artificial Intelligence and ML in simple words."
)
print(response.text)
```

```python
system_prompt="""
You are a helpful assistant.
Explain concepts in simple terms and provide clear, concise answers to questions.
keep answers in 100words or less.;
"""
question=input("Enter your question: ")
prompt=f"""
{system_prompt}

Student:
{question}
"""
response = client.models.generate_content(
    model="models/gemini-3.5-flash-lite",
    contents=prompt
)
print(response.text)
```

```python
system_prompt="""
You are my personal assistant. 
If someone asks about
politics,sports,movies,religion
reply 'I can only help with enginerring and academic topics.'
Never answer outside your domain.
"""

while True:
    question = input("\nYou:")

    if question.lower() == "exit":
        break
    prompt = f"""
{system_prompt}
Student: {question}"""


response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt
)
print("\nAssistant:")
print(response.text)
```
