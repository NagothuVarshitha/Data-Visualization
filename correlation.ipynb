```python
"""
study_hours + marks
       ↓
Scatter Plot
       ↓
Trend Line
       ↓
Correlation coefficient (r)
       ↓
Relationship between two numerical variables

r is close to +1, it is a strong positive relationship

close to -1, strong negative

close to 0, no linear relationship"""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("student_marks_with_study_hours_gender.csv")

print(df)
```

```python
#Select the two columns
X = df["StudyHours"]
Y = df["Python"]
```

```python
#reate the scatter plot
plt.figure(figsize=(7, 5))

sns.scatterplot(x=X, y=Y)

plt.xlabel("StudyHours")
plt.ylabel("Python")
plt.title("Study Hours vs Python")

plt.show()
```

```python
#Add the trend line
plt.figure(figsize=(7, 5))

sns.regplot(
    x=X,
    y=Y,
    scatter_kws={"s": 50},
    line_kws={"linestyle": "--"}
)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

```python
correlation = df["StudyHours"].corr(df["Python"])

print("Correlation:", correlation)
```

```python
correlation = df["StudyHours"].corr(df["Python"])

plt.figure(figsize=(7, 5))

sns.regplot(
    x=df["StudyHours"],
    y=df["Python"],
    scatter_kws={"s": 50},
    line_kws={"linestyle": "--"}
)

plt.title(f"StudyHours vs Python (r = {correlation:.2f})")
plt.xlabel("StudyHours")
plt.ylabel("Python")

plt.show()
```
