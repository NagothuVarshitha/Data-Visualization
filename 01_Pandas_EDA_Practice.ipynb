```python
df = pd.DataFrame({
    "name": ["A", "B", "C", "D"],
    "age": [20,None,22, None],
    "salary": [50000, 60000, 70000, None]
})
```

```python
"""step-1:count num of null values(.--dot means deeply diving into the data)
isnull-it will find the null values in the data
when we find null values we can use sum to get the total number of null values stores the total number of missing values in the entire DataFrame.df
 ↓
df.isnull()---How many missing values are in each column?
 ↓
True / False---How many missing values are there in total?
 ↓
.sum()
 ↓
x = missing values in EACH column
 ↓
x.sum()
 ↓
y = missing values in the WHOLE DataFrame
"""

```

```python
x=df.isnull().sum()
print(x)
```

```python
y=x.sum()
print(y)
```

```python
#total missing values in the dataframe
print("Total missing values in the dataframe:", df.isnull().sum().sum())
```

```python
customer = pd.DataFrame({
    "CustomerName": ["Varsha", "Geetha", "Karthik", "Javeed"],
    "CustomerID": [1, None, 3, 4],
    "CustomerPhoneNum": [1234567890, None, 3456789012, 4567890123],
    "CustomerEmail": [
        "varsha@example.com",
        None,
        "karthik@example.com",
        "javeed@example.com"
    ],
    "CustomerSalary": [50000, 60000, None, 80000],
    "CustomerAddress": ["123 Main St", "456 Elm St", None, "789 Oak St"]
})

print(customer)
```

```python
msno.matrix(df)
```

```python
customer.dropna()  #df.dropna() is used to remove rows that contain missing values (None / NaN).
```

```python
print(customer.dropna(how='all'))#------------------drop only rows that are entirely empty
```

```python
print(customer.dropna(subset=['CustomerID'])) #------------------drop rows where specific columns have missing values
```

```python
print(customer.dropna(thresh=2)) #------------------drop rows that have less than 3 non-missing values
```

```python
print(customer.dropna(subset=['CustomerID', 'CustomerPhoneNum'])) #------------------drop rows where specific columns have missing values
```

```python
customer.fillna(0) #------------------replace missing values with 0
```

```python
customer.ffill() #------------------forward fill missing values
```

```python
customer.bfill() #------------------backward fill missing values
```

```python
customer['CustomerSalary'].fillna(customer['CustomerSalary'].mean()) #------------------fill missing values with mean of the column
```

```python
customer['CustomerSalary'].fillna(customer['CustomerSalary'].mode()[0]) #------------------fill missing values with mode of the column
```

```python
customer['CustomerSalary'].fillna(customer['CustomerSalary'].median())#------------------fill missing values with median of the column
```

```python
student = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Bob", "David"],
    "branch": ["CS", "AIDS", "ECE", "AIDS", "BBA"],
    "cgpa": [3.5, 3.8, 3.2, 3.8, 3.9]
})

print(student)
```

```python
print(student.duplicated().sum()) #------------------check for duplicate rows
```

```python
student.drop_duplicates() #------------------drop duplicate rows
```

```python
student.drop_duplicates(keep='first') #------------------drop duplicate rows, keeping the first occurrence
```

```python
student.drop_duplicates(keep='last') #-----------------drop duplicate rows, keeping the last occurrence
```

```python
student.drop_duplicates(keep=False) #------------------drop all duplicate rows, keeping none of them
```
