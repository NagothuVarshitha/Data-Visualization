```python
import pandas as pd

students = pd.DataFrame({

    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],

    "Gender": ["Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female", "Male"],

    "Attendance": [92, 78, 88, 65, 95, 72, 85, 60, 90, 75,
                   98, 82, 87, 69, 93, 76, 89, 64, 96, 80],

    "Study_Hours": [8, 5, 7, 3, 9, 4, 6, 2, 8, 5,
                    10, 6, 7, 3, 9, 5, 8, 2, 10, 6],

    "Internal_Marks": [42, 35, 40, 28, 45, 32, 38, 25, 41, 34,
                       47, 39, 40, 29, 44, 36, 42, 27, 46, 37],

    "Assignment_Score": [18, 15, 17, 12, 19, 14, 16, 10, 18, 15,
                         20, 17, 18, 12, 19, 16, 18, 11, 20, 17],

    "Final_Marks": [88, 72, 84, 61, 94, 68, 79, 55, 87, 70,
                    97, 82, 85, 63, 92, 74, 89, 58, 96, 80]
})

print(students)
```

```python
#1. Attendance Distribution — Histogram
#We use a histogram because we want to see how the Attendance values are distributed.
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(students["Attendance"], bins=5, color="skyblue", edgecolor="black")

plt.title("Attendance Distribution")
plt.xlabel("Attendance (%)")
plt.ylabel("Number of Students")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
#2. Study Hours vs Final Marks — Scatter Plot
#Here we want to check whether more study hours are related to higher final marks.
plt.figure(figsize=(8, 5))

plt.scatter(
    students["Study_Hours"],
    students["Final_Marks"],
    color="green",
    marker="o"
)

plt.title("Study Hours vs Final Marks")
plt.xlabel("Study Hours")
plt.ylabel("Final Marks")

plt.grid(True, alpha=0.3)

plt.show()
```

```python
avg_marks = students.groupby("Gender")["Final_Marks"].mean()

print(avg_marks)
```

```python
#3. Average Marks by Gender — Bar Chart
#First, calculate the average final marks for each gender.
plt.figure(figsize=(7, 5))

plt.bar(
    avg_marks.index,
    avg_marks.values,
    color=["pink", "skyblue"]
)

plt.title("Average Final Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Marks")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
#4. Attendance vs Final Marks — Scatter Plot
#Now we check whether attendance is related to final marks.
plt.figure(figsize=(8, 5))

plt.scatter(
    students["Attendance"],
    students["Final_Marks"],
    color="purple",
    marker="o"
)

plt.title("Attendance vs Final Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Marks")

plt.grid(True, alpha=0.3)

plt.show()
```

```python
#5. Subplot — All 4 Graphs Together
#Now we put all four visualizations into one output.
fig, ax = plt.subplots(2, 2, figsize=(12, 9))

# 1. Histogram
ax[0, 0].hist(
    students["Attendance"],
    bins=5,
    color="skyblue",
    edgecolor="black"
)

ax[0, 0].set_title("Attendance Distribution")
ax[0, 0].set_xlabel("Attendance (%)")
ax[0, 0].set_ylabel("Number of Students")
ax[0, 0].grid(axis="y", alpha=0.3)


# 2. Study Hours vs Final Marks
ax[0, 1].scatter(
    students["Study_Hours"],
    students["Final_Marks"],
    color="green"
)

ax[0, 1].set_title("Study Hours vs Final Marks")
ax[0, 1].set_xlabel("Study Hours")
ax[0, 1].set_ylabel("Final Marks")
ax[0, 1].grid(True, alpha=0.3)


# 3. Average Marks by Gender
avg_marks = students.groupby("Gender")["Final_Marks"].mean()

ax[1, 0].bar(
    avg_marks.index,
    avg_marks.values,
    color=["pink", "skyblue"]
)

ax[1, 0].set_title("Average Final Marks by Gender")
ax[1, 0].set_xlabel("Gender")
ax[1, 0].set_ylabel("Average Final Marks")
ax[1, 0].grid(axis="y", alpha=0.3)


# 4. Attendance vs Final Marks
ax[1, 1].scatter(
    students["Attendance"],
    students["Final_Marks"],
    color="purple"
)

ax[1, 1].set_title("Attendance vs Final Marks")
ax[1, 1].set_xlabel("Attendance (%)")
ax[1, 1].set_ylabel("Final Marks")
ax[1, 1].grid(True, alpha=0.3)


plt.tight_layout()
plt.show()
```
