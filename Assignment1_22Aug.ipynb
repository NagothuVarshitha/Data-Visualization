```python
"""
Step 1: Import the libraries required for creating boxplots.
Seaborn is used for statistical visualization and Matplotlib is used to display the plots.
"""
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
"""
Step 2: Create boxplots for all seven numerical columns.
A boxplot helps us identify the median, quartiles, spread, and possible outliers.
"""
columns = ["Age","Unit_Price","Quantity","Discount","Rating","Delivery_Days","Total_Amount"]
fig,ax = plt.subplots(2,4,figsize=(16,8))
for i,column in enumerate(columns):
    sns.boxplot(y=df[column],ax=ax.flat[i])
    ax.flat[i].set_title(column)
    ax.flat[i].set_ylabel(column)
ax.flat[7].axis("off")
plt.tight_layout()
plt.show()

"""Observation:
The box represents the middle 50% of the data.
The line inside the box represents the median.
The whiskers show the normal range.
Individual points beyond the whiskers are possible outliers."""
```

```python
"""
Step 3: Calculate the number of possible outliers in each numerical column.
The IQR method identifies values below Q1 - 1.5*IQR or above Q3 + 1.5*IQR as possible outliers.
"""
for column in columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]
    print(column,":",len(outliers),"possible outliers")
```

```python
"""
Step 4: Identify which columns contain at least one possible outlier.
A column is considered to contain possible outliers when the IQR method finds one or more values outside its limits.
"""
for column in columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    outlier_count = ((df[column] < lower_limit) | (df[column] > upper_limit)).sum()
    if outlier_count > 0:
        print(column,":",outlier_count,"possible outliers")
```

```python
"""
Step 5: Find the column with the largest number of possible outliers.
The outlier counts are stored and the column with the highest count is identified.
"""
outlier_counts = {}
for column in columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    outlier_counts[column] = ((df[column] < lower_limit) | (df[column] > upper_limit)).sum()
print("Outlier counts:",outlier_counts)
print("Column with the largest number of possible outliers:",max(outlier_counts,key=outlier_counts.get))

"""Observation:
Total_Amount has the largest number of possible outliers:
2,472 possible outliers.
Unit_Price is next with 2,401."""
```

```python
"""**Task-11: Detect Outliers Using IQR**

For at least **three numerical columns**:

1. Calculate      Q .
2. Calculate      Q .
3. Calculate      IQR.
4. Calculate      the lower bound.
5. Calculate      the upper bound.
6. Identify      the outlier records.
7. Count      the number of outliers.

**Formula**
IQR = Q1-Q3 
Lower Bound = Q1-1.5×IQR
Upper Bound = Q3+1.5×IQR"""
```

```python
"""
Step 1: Calculate Q1, Q3, IQR, lower bound and upper bound for three numerical columns.
Q1 is the 25th percentile and Q3 is the 75th percentile.
IQR is calculated as Q3 minus Q1.
The lower and upper bounds are calculated using 1.5 times the IQR.
"""
columns = ["Age","Unit_Price","Quantity"]
for column in columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    print("\nColumn:",column)
    print("Q1:",q1)
    print("Q3:",q3)
    print("IQR:",iqr)
    print("Lower Bound:",lower_bound)
    print("Upper Bound:",upper_bound)
```

```python
"""
Step 2: Identify the records that fall outside the lower and upper bounds.
A value below the lower bound or above the upper bound is considered a possible outlier.
The complete records are displayed so that the unusual values can be investigated.
"""
for column in columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    print("\nOutlier Records for",column)
    print(outliers)
```

```python
"""
Step 3: Count the number of possible outliers in each selected numerical column.
The count is calculated by checking values below the lower bound or above the upper bound.
"""
for column in columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outlier_count = ((df[column] < lower_bound) | (df[column] > upper_bound)).sum()
    print(column,":",outlier_count,"outliers")
```

```python
"""
Step 1: Create a scatterplot to examine the relationship between customer Age and Total_Amount.
Each point represents one order.
"""

import pandas as pd  # Imports Pandas for data handling.
import seaborn as sns  # Imports Seaborn for visualization.
import matplotlib.pyplot as plt  # Imports Matplotlib for displaying graphs.

df = pd.read_csv("cleaned_ecommerce_dataset.csv")  # Loads the cleaned e-commerce dataset.

sns.scatterplot(data=df,x="Age",y="Total_Amount",color="#FFB6C1",s=60)  # Shows the relationship between Age and Total_Amount.
plt.title("Age vs Total Amount")  # Sets the graph title.
plt.xlabel("Age")  # Sets the x-axis label.
plt.ylabel("Total Amount")  # Sets the y-axis label.
plt.show()  # Displays the graph.

#Interpretation: There is no clear relationship between Age and Total_Amount, meaning customer age does not appear to strongly influence the order amount.
```

```python
"""
Step 2: Create a scatterplot to examine how Unit_Price is related to Total_Amount.
The plot helps identify whether higher product prices are associated with higher order amounts.
"""

sns.scatterplot(data=df,x="Unit_Price",y="Total_Amount",color="#FFB6C1",s=60)  # Shows the relationship between Unit_Price and Total_Amount.
plt.title("Unit Price vs Total Amount")  # Sets the graph title.
plt.xlabel("Unit Price")  # Sets the x-axis label.
plt.ylabel("Total Amount")  # Sets the y-axis label.
plt.show()  # Displays the graph.

#Interpretation: There is a strong positive relationship between Unit_Price and Total_Amount, so higher-priced products generally result in higher order amounts.
```

```python
"""
Step 3: Create a scatterplot to examine the relationship between Discount and Total_Amount.
This helps us understand whether discount percentage has a visible effect on order amount.
"""

sns.scatterplot(data=df,x="Discount",y="Total_Amount",color="#FFB6C1",s=60)  # Shows the relationship between Discount and Total_Amount.
plt.title("Discount vs Total Amount")  # Sets the graph title.
plt.xlabel("Discount (%)")  # Sets the x-axis label.
plt.ylabel("Total Amount")  # Sets the y-axis label.
plt.show()  # Displays the graph.

#Interpretation: There is no strong visible relationship between Discount and Total_Amount, so the discount percentage alone does not strongly determine the order amount.
```

```python
"""
Step 4: Create a boxplot to compare the distribution of Total_Amount across product categories.
The boxplot helps compare the median, spread and possible outliers for each category.
"""

sns.boxplot(data=df,x="Product_Category",y="Total_Amount",color="#FFB6C1")  # Compares Total_Amount distributions across product categories.
plt.title("Product Category vs Total Amount")  # Sets the graph title.
plt.xlabel("Product Category")  # Sets the x-axis label.
plt.ylabel("Total Amount")  # Sets the y-axis label.
plt.xticks(rotation=45)  # Rotates category names to improve readability.
plt.show()  # Displays the graph.

#Interpretation: Electronics shows much higher Total_Amount values than the other categories, while Books and Beauty generally have lower order amounts.
```

```python
"""
Step 5: Create a boxplot to compare Total_Amount across gender categories.
The boxplot helps compare the typical order amount and spread between the groups.
"""

sns.boxplot(data=df,x="Gender",y="Total_Amount",palette=["#FFB6C1","#ADD8E6","#FFB6C1"])  # Compares Total_Amount across the available gender categories.
plt.title("Gender vs Total Amount")  # Sets the graph title.
plt.xlabel("Gender")  # Sets the x-axis label.
plt.ylabel("Total Amount")  # Sets the y-axis label.
plt.show()  # Displays the graph.

#Interpretation: The order amounts vary across gender categories, but the distributions overlap considerably, so gender does not appear to be a strong factor in determining Total_Amount.
```

```python
"""
Step 6: Create a boxplot to compare Total_Amount across different payment methods.
The boxplot helps identify differences in order amount and possible extreme values between payment methods.
"""

sns.boxplot(data=df,x="Payment_Method",y="Total_Amount",color="#FFB6C1")  # Compares Total_Amount across payment methods.
plt.title("Payment Method vs Total Amount")  # Sets the graph title.
plt.xlabel("Payment Method")  # Sets the x-axis label.
plt.ylabel("Total Amount")  # Sets the y-axis label.
plt.xticks(rotation=45)  # Rotates payment method names for better readability.
plt.show()  # Displays the graph.

#Interpretation: The payment methods have broadly similar Total_Amount distributions, although some methods show slightly different medians and extreme values.
```
