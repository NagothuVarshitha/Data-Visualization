```python
import matplotlib.pyplot as plt     #Imports Matplotlib for creating graphs and charts.
import seaborn as sns      #Imports Seaborn for creating statistical and attractive visualizations.
import pandas as pd     #Imports Pandas for working with data and DataFrames.

df = pd.read_csv("student_marks.csv")     #Reads the student_marks.csv file and stores the data in a DataFrame called df.

print(df)    #Displays the contents of the DataFrame in the output.
```

```python
#bargraph
X = df["Name"]
Y = df["Student_ID"]
fig, ax = plt.subplots()
ax.plot(X, Y)
plt.show()
```

```python
#histogram
X = df["Python"]

fig, ax = plt.subplots()
ax.hist(X)

plt.show()
```

```python
# 1-Bar Graph
X = df["Name"]
Y = df["Python"]

fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(X, Y, color='blue', label='Python Marks')

ax.set_title("Python Marks of Students")
ax.set_xlabel("Student Name")
ax.set_ylabel("Python Marks")

plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
# 2-Pie Chart
subjects = ["Python", "SQL", "Maths", "English"]

marks = [
    df["Python"].mean(),
    df["SQL"].mean(),
    df["Maths"].mean(),
    df["English"].mean()
]

fig, ax = plt.subplots(figsize=(8, 6))

ax.pie(marks, labels=subjects, autopct='%1.1f%%')

ax.set_title("Average Marks by Subject")

plt.show()
```

```python
#3- Line Graph
X = df["Name"]
Y = df["Python"]

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(X, Y, marker='o', color='blue', label='Python Marks')

ax.set_title("Python Marks of Students")
ax.set_xlabel("Student Name")
ax.set_ylabel("Python Marks")

plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
# 4- Scatter Plot
X = df["Python"]
Y = df["SQL"]

fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(X, Y, marker='o', color='blue', label='Python vs SQL')

ax.set_title("Python vs SQL Marks")
ax.set_xlabel("Python Marks")
ax.set_ylabel("SQL Marks")

plt.legend()
plt.show()
```

```python
# 5-Histogram(each bar is called a bin)
X = df["Python"]

fig, ax = plt.subplots(figsize=(10, 6))

ax.hist(X, bins=5, color='blue', edgecolor='black')

ax.set_title("Distribution of Python Marks")
ax.set_xlabel("Python Marks")
ax.set_ylabel("Frequency")

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
df = pd.read_csv("student_marks.csv")

# Select columns
X = df["Name"]
Y = df["Python"]

# Create 2 × 3 dashboard
fig, ax = plt.subplots(2, 3, figsize=(15, 8))


# 1. Line Graph
ax[0, 0].plot(X, Y, marker='o', color='skyblue')
ax[0, 0].set_title("Python Marks - Line Graph")
ax[0, 0].set_xlabel("Student Name")
ax[0, 0].set_ylabel("Python Marks")
ax[0, 0].tick_params(axis='x', rotation=45)
ax[0, 0].grid(True, alpha=0.3)


# 2. Bar Graph
ax[0, 1].bar(X, Y, color='pink')
ax[0, 1].set_title("Python Marks - Bar Graph")
ax[0, 1].set_xlabel("Student Name")
ax[0, 1].set_ylabel("Python Marks")
ax[0, 1].tick_params(axis='x', rotation=45)
ax[0, 1].grid(axis='y', alpha=0.3)


# 3. Scatter Plot
ax[0, 2].scatter(
    df["Maths"],
    df["Python"],
    color='skyblue'
)
ax[0, 2].set_title("Maths vs Python Marks")
ax[0, 2].set_xlabel("Maths Marks")
ax[0, 2].set_ylabel("Python Marks")
ax[0, 2].grid(True, alpha=0.3)


# 4. Python Marks Histogram
ax[1, 0].hist(
    Y,
    bins=5,
    color='skyblue',
    edgecolor='black'
)
ax[1, 0].set_title("Python Marks Distribution")
ax[1, 0].set_xlabel("Python Marks")
ax[1, 0].set_ylabel("Frequency")
ax[1, 0].grid(axis='y', alpha=0.3)


# 5. SQL Marks Histogram
ax[1, 1].hist(
    df["SQL"],
    bins=5,
    color='pink',
    edgecolor='black'
)
ax[1, 1].set_title("SQL Marks Distribution")
ax[1, 1].set_xlabel("SQL Marks")
ax[1, 1].set_ylabel("Frequency")
ax[1, 1].grid(axis='y', alpha=0.3)


# 6. Pie Chart
ax[1, 2].pie(
    Y,
    labels=X,
    autopct='%1.1f%%',
    colors=['pink', 'skyblue'] * 5
)
ax[1, 2].set_title("Python Marks Distribution")


# Adjust layout
plt.tight_layout()

# Display dashboard
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("student_marks.csv")

X = df["Name"]
Y = df["Python"]

fig, ax = plt.subplots(figsize=(10, 6))

colors = ["#FF5733", "#33FF57", "#3357FF", "#F333FF"]

for i in range(len(df)):
    ax.scatter(X[i], Y[i], color=colors[i % 4], label=X[i])

ax.legend(title="Students")
ax.grid(alpha=0.3)

plt.xlabel("Student Name")
plt.ylabel("Python Marks")
plt.title("Python Marks by Student")

plt.savefig("python_marks.png")
plt.show()
```

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV file
df = pd.read_csv("messy_ecommerce_15000_student_practice.csv")

# Display dataset dimensions
print("Dataset Dimensions:", df.shape)

# Display number of rows
print("Number of Rows:", df.shape[0])

# Display number of columns
print("Number of Columns:", df.shape[1])

# Display column names
print("Names of Columns:", df.columns.tolist())

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display dataset information
print("\nData Info:")
df.info()

# Display Order_Date data type
print("\nOrder Date Data Type:", df["Order_Date"].dtype)

# Display missing values in each column
print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# Visualize missing values
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False)

# Add title
plt.title("Missing Values in E-Commerce Dataset")

# Display graph
plt.show()

# Count duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Display duplicate records
print("\nDuplicate Rows:")
print(df[df.duplicated()])

# Count duplicate Order IDs
print("\nDuplicate Order IDs:", df["Order_ID"].duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Display new dataset shape
print("\nDataset Shape After Removing Duplicates:", df.shape)
```
